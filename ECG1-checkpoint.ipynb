```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, GlobalAveragePooling1D, Dropout, Reshape, SeparableConv2D,
    BatchNormalization, ReLU, Add, LayerNormalization, MultiHeadAttention
)
from tensorflow.keras.initializers import glorot_uniform
from tensorflow.keras.callbacks import ModelCheckpoint, TerminateOnNaN, EarlyStopping
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.model_selection import train_test_split
from PIL import Image
import matplotlib.pyplot as plt

# Parameters
img_size = (224, 224)
batch_size = 4
data_dir = r"D:\ECG"

# Filter label folders
label_names = sorted([d for d in os.listdir(data_dir) if os.path.isdir(os.path.join(data_dir, d))])
label_map = {name: idx for idx, name in enumerate(label_names)}

# Load image paths and labels
image_paths, labels = [], []
for label_name in label_names:
    class_path = os.path.join(data_dir, label_name)
    for img_file in tf.io.gfile.glob(os.path.join(class_path, "*.png")):
        image_paths.append(img_file)
        labels.append(label_map[label_name])

# Convert to numpy array
image_paths = np.array(image_paths)
labels = np.array(labels)

# Split dataset
train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    image_paths, labels, test_size=0.2, stratify=labels, random_state=42)
val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths, temp_labels, test_size=0.5, stratify=temp_labels, random_state=42)

# Data generator
def create_generator(image_paths, labels, augment=False):
    def generator():
        for path, label in zip(image_paths, labels):
            img = Image.open(path).resize(img_size).convert('RGB')
            img = np.array(img) / 255.0
            yield img, tf.keras.utils.to_categorical(label, num_classes=len(label_names))

    return tf.data.Dataset.from_generator(
        generator,
        output_signature=(
            tf.TensorSpec(shape=(img_size[0], img_size[1], 3), dtype=tf.float32),
            tf.TensorSpec(shape=(len(label_names),), dtype=tf.float32)
        )
    ).shuffle(1024).batch(batch_size).prefetch(tf.data.AUTOTUNE)

train_ds = create_generator(train_paths, train_labels, augment=True)
val_ds = create_generator(val_paths, val_labels)
test_ds = create_generator(test_paths, test_labels)

# --- Vision Transformer Modules ---
def patch_embedding(x, patch_size=4, embed_dim=128):
    x = SeparableConv2D(embed_dim, (patch_size, patch_size), strides=4, padding='same',
                        kernel_initializer=glorot_uniform())(x)
    x = Reshape((-1, embed_dim))(x)
    return x

def apply_rope(x):
    seq_len = tf.shape(x)[1]
    dim = tf.shape(x)[2]
    pos = tf.range(seq_len, dtype=tf.float32)
    freq = tf.range(0, dim, 2.0, dtype=tf.float32) / tf.cast(dim, tf.float32)
    freq = tf.exp(-freq * tf.math.log(10000.0))
    angles = tf.einsum('i,j->ij', pos, freq)
    rope = tf.concat([tf.sin(angles), tf.cos(angles)], axis=-1)
    return x + rope

class GatedMultiHeadSelfAttention(layers.Layer):
    def __init__(self, num_heads=4, key_dim=64, embed_dim=128, **kwargs):
        super(GatedMultiHeadSelfAttention, self).__init__(**kwargs)
        self.num_heads = num_heads
        self.key_dim = key_dim
        self.embed_dim = embed_dim
        self.mhsa = MultiHeadAttention(num_heads=num_heads, key_dim=key_dim)
        self.gate_dense = Dense(embed_dim, activation='sigmoid')

    def call(self, x):
        attn_output = self.mhsa(x, x)
        gate = self.gate_dense(x)
        gated_attn = attn_output * gate
        return Add()([x, gated_attn])

    def get_config(self):
        config = super().get_config()
        config.update({
            "num_heads": self.num_heads,
            "key_dim": self.key_dim,
            "embed_dim": self.embed_dim,
        })
        return config

def ds_ffn(x, expansion_factor=4):
    dim = x.shape[-1]
    x = Dense(dim * expansion_factor, activation='gelu')(x)
    x = Dense(dim)(x)
    return x

def stochastic_depth(x, drop_prob=0.1):
    return Dropout(rate=drop_prob)(x) if drop_prob > 0 else x

def build_model(input_shape, num_transformer_layers=4, num_heads=4, key_dim=64, embed_dim=128):
    inputs = Input(shape=input_shape)
    x = patch_embedding(inputs)
    x = apply_rope(x)
    for _ in range(num_transformer_layers):
        x = LayerNormalization()(x)
        x = GatedMultiHeadSelfAttention(num_heads=num_heads, key_dim=key_dim, embed_dim=embed_dim)(x)
        x = LayerNormalization()(x)
        x = ds_ffn(x)
        x = stochastic_depth(x)
    x = GlobalAveragePooling1D()(x)
    outputs = Dense(len(label_names), activation='softmax')(x)
    return Model(inputs, outputs)

# Build & compile model
model = build_model((*img_size, 3))
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Callbacks
model_save_dir = 'C:/Users/VARUNPGOPI/SRIVALLI/MODELS/'
os.makedirs(model_save_dir, exist_ok=True)

checkpoint_cb = ModelCheckpoint(
    os.path.join(model_save_dir, 'SRIVALLI_001_8_ES_20_val_loss_{val_loss:.4f}.h5'),
    save_best_only=True, monitor='val_loss', mode='min', verbose=1)
early_stopping_cb = EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True, verbose=1)
terminate_on_nan = TerminateOnNaN()

# Train model
history = model.fit(train_ds, validation_data=val_ds, epochs=1000, callbacks=[checkpoint_cb, terminate_on_nan, early_stopping_cb])

# Plot Accuracy and Loss
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train')
plt.plot(history.history['val_accuracy'], label='Val')
plt.title('Model Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Val')
plt.title('Model Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend()
plt.tight_layout(); plt.show()
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score, accuracy_score
import numpy as np
import tensorflow as tf

# Load best model
model = tf.keras.models.load_model("best_ecg_model.h5")

# Prepare test dataset again
test_ds = create_generator(test_paths, test_labels, augment=False)

# Get predictions
y_true = []
y_pred = []

for images, labels in test_ds:
    preds = model.predict(images)
    y_true.extend(np.argmax(labels.numpy(), axis=1))
    y_pred.extend(np.argmax(preds, axis=1))

y_true = np.array(y_true)
y_pred = np.array(y_pred)

# Metrics
acc = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average='macro')
recall = recall_score(y_true, y_pred, average='macro')
f1 = f1_score(y_true, y_pred, average='macro')
cm = confusion_matrix(y_true, y_pred)

# Dice Score (per class)
def dice_score(conf_matrix):
    dice_per_class = []
    for i in range(conf_matrix.shape[0]):
        TP = conf_matrix[i, i]
        FP = np.sum(conf_matrix[:, i]) - TP
        FN = np.sum(conf_matrix[i, :]) - TP
        dice = (2 * TP) / (2 * TP + FP + FN + 1e-7)
        dice_per_class.append(dice)
    return np.mean(dice_per_class), dice_per_class

# IoU Score (per class)
def iou_score(conf_matrix):
    iou_per_class = []
    for i in range(conf_matrix.shape[0]):
        TP = conf_matrix[i, i]
        FP = np.sum(conf_matrix[:, i]) - TP
        FN = np.sum(conf_matrix[i, :]) - TP
        iou = TP / (TP + FP + FN + 1e-7)
        iou_per_class.append(iou)
    return np.mean(iou_per_class), iou_per_class

dice_avg, dice_all = dice_score(cm)
iou_avg, iou_all = iou_score(cm)

print("\n📊 **Test Metrics**")
print(f"Accuracy        : {acc:.4f}")
print(f"Precision (macro): {precision:.4f}")
print(f"Recall (Sens.)   : {recall:.4f}")
print(f"F1 Score         : {f1:.4f}")
print(f"Mean Dice Score  : {dice_avg:.4f}")
print(f"Mean IoU Score   : {iou_avg:.4f}")

# Sensitivity (Recall) and Specificity per class
TP = np.diag(cm)
FN = np.sum(cm, axis=1) - TP
FP = np.sum(cm, axis=0) - TP
TN = np.sum(cm) - (TP + FP + FN)

sensitivity_per_class = TP / (TP + FN + 1e-7)
specificity_per_class = TN / (TN + FP + 1e-7)

print("\nClass-wise Sensitivity:", np.round(sensitivity_per_class, 4))
print("Class-wise Specificity:", np.round(specificity_per_class, 4))
```
